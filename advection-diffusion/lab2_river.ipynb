```python
D = 1
U = 1
T = 10*3600 * si.s
L = 1295 * si.m
```

```python
N = 1000
n = 1000
dt = T/n
dx = L/N
t = np.linspace(0, T, n+1)

c = c_new = np.zeros((N+1))
c[4] = 1
Ca = U * dt / dx
Cd = D * dt /dx**2

for j in range(2,N):
    a1 = (Cd*(1-Ca) - Ca/6 * (Ca**2-3*Ca+2))
    a2 = -Cd*(2-3*Ca)+Ca/2*(Ca**2-2*Ca-1)
    a3 = Cd*(1-3*Ca)-Ca/2*(Ca**2-Ca-2)
    a4 = Cd*Ca+Ca/6*(Ca**2-1)
    c_new[j]=c[j] + a1*c[j+1] + a2*c[j] + a3*c[j-1] + a4*c[j-2]
```

```python
plt.plot(t, c)
```
