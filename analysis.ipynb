# Metoda
---
1. **Jakie warunki rozważamy?** Dane wejściowe do modelu (fizyczne z jednostkami).
2. **Jakiej metody numerycznej używamy** Krótkie omówienie metody - podanie wzoru, przyjęta dyskretyzacja zjawiska, kroki czasowe i przestrzenne, liczba kroków, ...

```python
# wartości rzeczywiste
def my_simple_function(n, l, h, x_0):
    x = np.zeros(n)
    x[0] = x_0
    for i in range(1, n):
        x[i] = x[i-1]*(l*h+1) 
    return x
```

# Wyniki i analiza
--- 
Do analizowania modeli numerycznych możemy wykorzystywać szereg narzędzi i metod. Im bardziej złożona metoda, tym ciekawsze przypadki testów. Dzięki wspomnianemu wcześniej Pytestowi możemy prześledzić poprawność implementacji pod różnym kątem (sprawdzalność wyników, wyniki dla różnych warunków początkowych i brzegowych, innych kroków czasowych i przestrzennych, ...). Możliwości jest bardzo dużo.

Ale! Dzięki temu, że zajmujemy się zjawiskami fizycznymi testy mamy od razu (prawie zawsze:)). 

## Pytest - in a seperate file
[AAAC](https://docs.pytest.org/en/7.1.x/explanation/anatomy.html)

```python
@pytest.fixture
@pytest.mark.parametrize(
    "n, x_0, expected", (
            (10, 0.0, 0.0),
            (100, 0.0, 0.0 ),
            (1, 1.0, 1.0)
    )
)
def test_simple_function(n, x_0, expected):
    # Arrange
    time_step = 1
    x_step = 1
    sut=my_simple_function
    
    # Act
    actual=sut(
        n=n,
        x_0=x_0,
        l=x_step,
        h=time_step,
    )
    
    #Assert
    np.testing.assert_almost_equal(actual, expected)
    
```

```python
actual = my_simple_function(n=10, x_0=0, l=1, h=1)
np.testing.assert_equal(actual, 1)
```
